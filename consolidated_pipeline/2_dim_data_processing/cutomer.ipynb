```python
df_bronze = spark.sql(
    f"""
    SELECT * 
    FROM {catalog}.{bronze_schema}.{data_source}
    """
)
df_bronze.show(10)
```

### Transformation

```python
df_duplicates = (
    df_bronze
    .groupBy("customer_id")
    .count()
    .filter(F.col("count") > 1)
)
# df_duplicates.show()

# df_duplicates = (
#     df_bronze
#     .groupBy("customer_id")
#     .agg(F.count("customer_id"))
#     .where("count(customer_id) > 1")
# )
# df_duplicates.show()

print('Rows before duplicates dropped: ', df_bronze.count())
df_silver = df_bronze.dropDuplicates(['customer_id'])
print('Rows after duplicates dropped: ', df_silver.count())
```

```python
df_trim = (
    df_silver
    .withColumn(
        "customer_name_trim",
        F.trim(F.col("customer_name"))
    )
    .where("customer_name_trim != customer_name")
    .select("customer_name",  "customer_name_trim")
)

df_trim.show(truncate=False)
```

```python
df_silver = (
    df_silver
    .withColumn(
        "customer_name",
        F.trim(F.col("customer_name"))
    )
)
```

```python
df_silver.show()
```

```python
# df_silver.select('city').distinct().show()
city_mapping = {
    'Bengaluruu': 'Bengaluru',
    'Bengalore': 'Bengaluru',

    'Hyderabadd': 'Hyderabad',
    'Hyderbad': 'Hyderabad',

    'NewDelhi': 'New Delhi',
    'NewDheli': 'New Delhi',
    'NewDelhee': 'New Delhi'
}


allowed = ["Bengaluru", "Hyderabad", "New Delhi"]

df_silver = (
    df_silver
    .replace(city_mapping, subset=["city"])
    .withColumn(
        "city",
        F.when(F.col("city").isNull(), None)
         .when(F.col("city").isin(allowed), F.col("city"))
         .otherwise(None)
    )
)

df_silver.select('city').distinct().show()
```
